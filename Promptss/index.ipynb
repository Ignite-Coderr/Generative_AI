```python
# 1. Zero-shot prompting :- Direct prompting without providing any examples to the model.
# 2. Few-Shot prompting :- Providing a few examples to the model in the prompt to help 
#                          it understand the task better.

# 3. Chain of thought :- Providing a step-by-step reasoning process in the prompt to help the model 
#                        arrive at the correct answer.

# 4. Self-consistency :- The model generates multiple answers to the same question and then selects 
#                        the most consistent answer among them.

# 5. Persona-based prompting :- Providing a specific persona or character for the model to adopt in the
#                               prompt,which can help in generating more contextually relevant responses.

# 6. Role-playing :- Assigning specific roles to the model in the prompt, such as a teacher, student, 
#                    or interviewer, to guide the response generation.

# 7. Contextual prompting :- Providing relevant context or background information in the prompt to help the model
#                            generate more accurate and relevant responses.

# 8. Multimodeal prompting :- Incorporating multiple modalities, such as text, images, or audio, in the 
#                             prompt to enhance the model's understanding and response generation.
```
