```python
import requests

def generate_completion():
    url = "https://api.euron.one/api/v1/euri/chat/completions"
    headers = {
        "Content-Type": "application/json",
        "Authorization": "Bearer your_euri_api_token_here"
    }
    payload = {
        "messages": [
            {
                "role": "user",
                "content": "Who is the current PM of India?"
            }
        ],
        "model": "gpt-4.1-nano",
        "max_tokens": 1000,
        "temperature": 0.7
    }

    response = requests.post(url, headers=headers, json=payload)
    data = response.json()
    print(data)

generate_completion()
```

```python
import requests
import json

# Your API token
api_token = "YOUR_API_TOKEN"

# Define the API endpoint
url = "https://api.euron.one/api/v1/euri/images/generations"

# Define the headers
headers = {
    "Content-Type": "application/json",
    "Authorization": "Bearer your_euri_api_token_here"
}

# Define the data payload
data = {
    "prompt": "Kissing Couple",
    "model": "black-forest-labs/FLUX.1-schnell",
    "n": 2,
    "size": "1024x1024",
    "quality": "standard",
    "response_format": "url",
    "style": "vivid"
}

# Make the POST request
response = requests.post(url, headers=headers, data=json.dumps(data))

# Check the response status and print the result
if response.status_code == 200:
    print("Response JSON:", response.json())
else:
    print(f"Error: {response.status_code}, {response.text}")
```

```python
# Text to speech
import requests
import json

# Your API token
api_token = "YOUR_API_TOKEN"

# Define the API endpoint
url = "https://api.euron.one/api/v1/euri/audio/speech"

# Define the headers
headers = {
    "Content-Type": "application/json",
    "Authorization": "Bearer your_euri_api_token_here"
}

# Define the data payload
data = {
    "input": "Hello, this is a text message",
    "model": "playai-tts",
    "voice": "Fritz-PlayAI"
}

# Make the POST request
response = requests.post(url, headers=headers, data=json.dumps(data))

# Check the response status and print the result
if response.status_code == 200:
    print("Raw response:", response.text)
else:
    print(f"Error: {response.status_code}, {response.text}")
```
