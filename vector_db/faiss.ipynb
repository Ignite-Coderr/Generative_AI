```python
import numpy as np
import os
import requests
import json
```

```python
def chunk_text(text, chunk_size=50, overlaap=5):
    # Split the input text into a list of individual words
    words = text.split()
    
    # Initialize an empty list to store the text chunks
    chunks = []
    
    # Start index for chunking
    start = 0
    
    # Loop until we have processed all words
    while start < len(words):
        
        # Calculate the end index of the current chunk
        # Ensures we do not go beyond the total number of words
        end = min(start + chunk_size, len(words))
        
        # Join the words from start to end into a single text chunk
        chunk = ' '.join(words[start:end])
        
        # Add the generated chunk to the chunks list
        chunks.append(chunk)
        
        # Move the start index forward while keeping overlap between chunks
        start += chunk_size - overlaap
    
    # Return the list of overlapping text chunks
    return chunks
```

```python
header = {
    "Content-Type": "application/json",
    "Authorization": f"Bearer {API_KEY}"
}
```

```python
# Converting data into embeddings through chunks
all_embeddings = []
for i ,chunk in enumerate(chunks):
    payload = {
        "model": MODEL_NAME,
        "input": chunk
    }
    response = requests.post(API_URL, headers=header, data=json.dumps(payload))
    data = response.json()
    embeddings = data['data'][0]['embedding']
    all_embeddings.append(embeddings)
    
    
```

1️⃣ faiss.IndexFlatL2

This creates a FAISS index.

- FAISS = Facebook AI Similarity Search
- Used for fast similarity search on vectors (embeddings)

IndexFlatL2 specifically means:

- A flat (brute-force) index using L2 (Euclidean) distance

2️⃣ What does “Flat” mean?

- Flat = no clustering, no compression, no approximation
- FAISS will compare the query vector with every stored vector

```python
base_index = faiss.IndexFlatL2(1536)    # L2 :- Euclidean Distance
```

```python
# Storing the data
base_index.add(embeddings_array)
```

```python
# Creating db(faiss) and storing data into it
faiss.write_index(base_index,"faiss_index.faiss")
```

```python
def embedding_text(text):
    payload = {
        "model": MODEL_NAME,
        "input": text
    }
    response = requests.post(API_URL,headers=header,data = json.dumps(payload))
    data = response.json()
    embeddings = data['data'][0]['embedding']

    # After converting into embedding converting data into float32 format
    emb = np.array(embeddings,dtype='float32').reshape(1,-1)
    return emb
```
